1. Initialize a Spark Session and and Context

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("spark-chap1-sol")
    .master("local[4]")
    .config("spark.driver.memory", "4g")
    .config("spark.sql.shuffle.partitions", "8")
    .getOrCreate()
)
sc = spark.sparkContext
```

2. Download the data from this [link](https://raw.githubusercontent.com/inesmcm26/lp-big-data/main/data/people-100.csv) and save it. The data is a CSV file with information about people.

```python
import urllib.request

urllib.request.urlretrieve('https://raw.githubusercontent.com/inesmcm26/lp-big-data/main/data/people-100.csv', 'people-100.csv')
```

3. View the contents of the file using SQL

```python
people_df = spark.read.csv("people-100.csv", header=True)
people_df.createOrReplaceTempView("people")
```

```python
spark.sql("SELECT * FROM people").show()
```

4. Find out how many people work as `Retail banker` using SQL

```python
spark.sql(
"""
SELECT count(*)
FROM people
WHERE `Job Title` = 'Retail banker';
"""
).show()
```

5. Find out how many males and females are in the dataset using SQL

```python
spark.sql(
"""
SELECT count(*)
FROM people
GROUP BY `Gender`;
"""
).show()
```

6. Create an RDD containing integers from 1 to 10

```python
data = range(1, 11)

rdd = sc.parallelize(data)
```

7. Calculate the sum of the elements in the RDD using an action

Hint: Check out all RDD methods [here](https://spark.apache.org/docs/latest/api/python/reference/api/pyspark.RDD.html#pyspark.RDD)

```python
rdd.reduce(lambda x, y: x + y)
```

8. Filter the RDD to keep only the even numbers and return the result using a transformation and an action

```python
rdd.filter(lambda x: x % 2 == 0)
```

9. Apply the `map` transformation to the following RDD to create the paired RDD where the keys are the values in the original RDD and the values are random numbers from 1 to 10.

```python
data = ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C']

rdd = sc.parallelize(data)
```

```python
paired_rdd = rdd.map(lambda x: (x, random.randint(1, 11)))

paired_rdd.collect()
```

10. Find the maximum value for each key in the Paired RDD

```python
max_per_key = paired_rdd.reduceByKey(lambda x, y: max(x, y))

max_per_key.collect()
```

11. Calculate the sum of the values for each key in the Paired RDD

```python
sum_per_key = paired_rdd.reduceByKey(lambda x, y: x + y)
```
